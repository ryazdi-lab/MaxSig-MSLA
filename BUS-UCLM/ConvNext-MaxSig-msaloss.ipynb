```python
# %%
import os
from pathlib import Path

import cv2
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from time import perf_counter

# %%
# =========================================================
# 0) Utils: Torch Dice / IoU (works with logits)
# =========================================================

@torch.no_grad()
def dice_from_logits(logits: torch.Tensor, targets: torch.Tensor, threshold: float = 0.5, eps: float = 1e-6) -> torch.Tensor:
    """
    logits:  [B, 1, H, W] (raw, no sigmoid)
    targets: [B, 1, H, W] float {0,1}
    """
    probs = torch.sigmoid(logits)
    preds = (probs > threshold).float()
    dims = (1, 2, 3)
    intersection = (preds * targets).sum(dims)
    denom = preds.sum(dims) + targets.sum(dims)
    dice = (2.0 * intersection + eps) / (denom + eps)
    return dice.mean()

@torch.no_grad()
def iou_from_logits(logits: torch.Tensor, targets: torch.Tensor, threshold: float = 0.5, eps: float = 1e-6) -> torch.Tensor:
    probs = torch.sigmoid(logits)
    preds = (probs > threshold).float()
    dims = (1, 2, 3)
    intersection = (preds * targets).sum(dims)
    union = preds.sum(dims) + targets.sum(dims) - intersection
    iou = (intersection + eps) / (union + eps)
    return iou.mean()

@torch.no_grad()
def evaluate_model(model: nn.Module, loader: DataLoader, device: str = "cuda") -> dict:
    model.eval()
    criterion = nn.BCEWithLogitsLoss()

    total_loss = 0.0
    total_dice = 0.0
    total_iou = 0.0
    n = 0

    for images, masks in loader:
        images = images.to(device)
        masks = masks.to(device).float().unsqueeze(1)  # [B,1,H,W]

        logits = model(images)  # [B,1,h,w]
        logits = F.interpolate(logits, size=masks.shape[-2:], mode="bilinear", align_corners=False)

        loss = criterion(logits, masks).item()
        d = dice_from_logits(logits, masks).item()
        j = iou_from_logits(logits, masks).item()

        total_loss += loss
        total_dice += d
        total_iou += j
        n += 1

    if n == 0:
        return {"loss": 0.0, "dice": 0.0, "iou": 0.0}

    return {"loss": total_loss / n, "dice": total_dice / n, "iou": total_iou / n}

# %%
# ============================================================
# MaxSig-Aware Loss (MSAL)
# ============================================================
class MaxSigFeatureHook:
    """Collects MaxSigLayerNormalized outputs during forward pass."""
    def __init__(self, model):
        self.features = {}
        self.handles = []
        for name, module in model.named_modules():
            if isinstance(module, MaxSigLayerNormalized):
                self.handles.append(
                    module.register_forward_hook(self._make_hook(name))
                )

    def _make_hook(self, name):
        def hook(_, __, output):
            self.features[name] = output
        return hook

    def clear(self):
        self.features.clear()

    def remove(self):
        for h in self.handles:
            h.remove()


def _local_stats(x, k=3):
    """Return (center, mean, median) over kxk patches. x: (B,C,H,W)."""
    B, C, H, W = x.shape
    pad = k // 2
    uf = F.unfold(x, kernel_size=k, padding=pad)              # (B, C*k*k, L)
    uf = uf.view(B, C, k * k, H, W)
    center = uf[:, :, k * k // 2, :, :]
    mean_l = uf.mean(dim=2)
    med_l, _ = uf.median(dim=2)
    return center, mean_l, med_l


class MaxSigAwareLoss(nn.Module):
    """
    L = a_bce * BCE + a_dice * Dice
        + a_sas * SAS
        + a_mbe * MBE
        + a_mmd * MMD
        + a_pdr * PDR(model)
    """
    def __init__(self,
                 alpha_bce=1.0,
                 alpha_dice=1.0,
                 alpha_sas=0.05,  # Reduced from 0.5 to prevent gradient explosion
                 alpha_mbe=0.0,   # Set to 0.0 to avoid double-counting BCE
                 alpha_mmd=0.05,  # Reduced from 0.2
                 alpha_pdr=1e-3,
                 k=3, eps=1e-7):
        super().__init__()
        self.a_bce   = alpha_bce
        self.a_dice  = alpha_dice
        self.a_sas   = alpha_sas
        self.a_mbe   = alpha_mbe
        self.a_mmd   = alpha_mmd
        self.a_pdr   = alpha_pdr
        self.k = k
        self.eps = eps
        self.bce = nn.BCEWithLogitsLoss(reduction='none')

    # ---- term 1: Significance-Aligned Segmentation ----
    def sas_loss(self, probs, targets):
        pc, pm, pmd = _local_stats(probs, self.k)
        tc, tm, tmd = _local_stats(targets, self.k)
        sig_p = (pc - pm).abs() + (pc - pmd).abs()
        sig_t = (tc - tm).abs() + (tc - tmd).abs()
        # Detach amax to prevent gradient explosion through normalization
        sig_p = sig_p / (sig_p.amax(dim=(2, 3), keepdim=True).detach() + self.eps)
        sig_t = sig_t / (sig_t.amax(dim=(2, 3), keepdim=True).detach() + self.eps)
        return F.l1_loss(sig_p, sig_t)

    # ---- term 2: Max-Bounded Boundary Emphasis ----
    def mbe_loss(self, logits, targets):
        bce = self.bce(logits, targets)                       # (B,1,H,W)
        pc, pm, _ = _local_stats(torch.sigmoid(logits), self.k)
        tc, tm, _ = _local_stats(targets, self.k)
        bnd_p = (pc - pm).abs()
        bnd_t = (tc - tm).abs()
        w = torch.maximum(bnd_p, bnd_t)
        w = w / (w.mean() + self.eps)
        return (bce * w).mean()

    # ---- term 3: Median-Mean Divergence ----
    def mmd_loss(self, probs, targets):
        _, pm, pmd = _local_stats(probs, self.k)
        _, tm, tmd = _local_stats(targets, self.k)
        div_p = (pm - pmd).abs()
        div_t = (tm - tmd).abs()
        return F.l1_loss(div_p, div_t)

    # ---- term 4: Parameter Diversity Regularization ----
    def pdr_loss(self, model):
        reg = None
        for m in model.modules():
            if isinstance(m, MaxSigLayerNormalized):
                w = torch.sigmoid(m.weight).clamp(self.eps, 1 - self.eps)
                p = w / (w.sum() + self.eps)
                ent = -(p * p.log()).sum()
                pos_pen = F.relu(-m.weight_center) + F.relu(-m.weight_median)
                bal = (m.weight_center - m.weight_median).abs()
                term = -ent + pos_pen + 0.1 * bal
                reg = term if reg is None else reg + term
        return reg if reg is not None else torch.tensor(0.0, device=next(model.parameters()).device)

    def dice_loss(self, probs, targets):
        inter = (probs * targets).sum(dim=(2, 3))
        union = (probs + targets).sum(dim=(2, 3))
        dice = (2 * inter + self.eps) / (union + self.eps)
        return 1 - dice.mean()

    def forward(self, logits, targets, model=None):
        probs = torch.sigmoid(logits)
        loss  = self.a_bce  * self.bce(logits, targets).mean()
        loss  = loss + self.a_dice * self.dice_loss(probs, targets)
        
        # Safely add auxiliary terms if their weights are active
        if self.a_sas > 0:
            loss = loss + self.a_sas * self.sas_loss(probs, targets)
        if self.a_mbe > 0:
            loss = loss + self.a_mbe * self.mbe_loss(logits, targets)
        if self.a_mmd > 0:
            loss = loss + self.a_mmd * self.mmd_loss(probs, targets)
        if model is not None and self.a_pdr > 0:
            loss = loss + self.a_pdr * self.pdr_loss(model)
            
        return loss

# %%

# ============================================================
# MaxSigLayerNormalized (with proper C_in -> C_out projection)
# ============================================================
class MaxSigLayerNormalized(nn.Module):
    def __init__(
        self,
        ch_in: int,
        ch_out: int,
        kernel_size: int,
        stride: int = 1,
        use_sigmoid_input: bool = True,
    ):
        super().__init__()
        assert kernel_size >= 1
        self.ch_in = ch_in
        self.ch_out = ch_out
        self.kernel_size = kernel_size
        self.stride = stride
        self.use_sigmoid_input = use_sigmoid_input

        self.padding = (kernel_size - 1) // 2
        self.kernel_area = kernel_size ** 2
        self.center_index = self.kernel_area // 2

        self.weight = nn.Parameter(torch.rand(1, 1, self.kernel_area, 1, 1))
        self.bias   = nn.Parameter(torch.zeros(1, 1, self.kernel_area, 1, 1))

        self.weight_center = nn.Parameter(torch.tensor(1.0))
        self.weight_median = nn.Parameter(torch.tensor(1.0))

        self.proj = nn.Conv2d(ch_in, ch_out, kernel_size=1, bias=False)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        B, C_in, H, W = x.shape
        assert C_in == self.ch_in, f"Expected {self.ch_in} channels, got {C_in}"

        x_in = torch.sigmoid(x) if self.use_sigmoid_input else x

        x_unfold = F.unfold(
            x_in,
            kernel_size=self.kernel_size,
            padding=self.padding,
            stride=self.stride,
        )  # (B, C_in*K*K, L)

        _, CK2, L = x_unfold.shape
        assert CK2 == C_in * self.kernel_area

        out_h = (H + 2 * self.padding - self.kernel_size) // self.stride + 1
        out_w = (W + 2 * self.padding - self.kernel_size) // self.stride + 1

        x_unfold = x_unfold.view(B, C_in, self.kernel_area, out_h, out_w)

        w = torch.sigmoid(self.weight)
        b = torch.sigmoid(self.bias)

        D = torch.maximum(w, x_unfold) + b

        mean_stat = torch.mean(D, dim=2)
        median_stat, _ = torch.median(D, dim=2)

        center_values = x_unfold[:, :, self.center_index, :, :]

        out = (
            self.weight_center * center_values
            + self.weight_median * median_stat
            - mean_stat
        )

        out = self.proj(out)
        return out


# %%
# =========================================================
# 1) ConvNeXt Building Blocks
# =========================================================

class LayerNorm(nn.Module):
    def __init__(self, normalized_shape, eps=1e-6, data_format="channels_last"):
        super().__init__()
        self.weight = nn.Parameter(torch.ones(normalized_shape))
        self.bias = nn.Parameter(torch.zeros(normalized_shape))
        self.eps = eps
        if data_format not in ("channels_last", "channels_first"):
            raise ValueError("data_format must be 'channels_last' or 'channels_first'")
        self.data_format = data_format
        self.normalized_shape = (normalized_shape,)

    def forward(self, x):
        if self.data_format == "channels_last":
            mean = x.mean(-1, keepdim=True)
            var = (x - mean).pow(2).mean(-1, keepdim=True)
            x = (x - mean) / torch.sqrt(var + self.eps)
            x = x * self.weight + self.bias
            return x
        else:
            mean = x.mean(1, keepdim=True)
            var = (x - mean).pow(2).mean(1, keepdim=True)
            x = (x - mean) / torch.sqrt(var + self.eps)
            x = x * self.weight[:, None, None] + self.bias[:, None, None]
            return x

class DropPath(nn.Module):
    def __init__(self, drop_prob=0.):
        super().__init__()
        self.drop_prob = drop_prob

    def forward(self, x):
        if self.drop_prob == 0. or not self.training:
            return x
        keep_prob = 1 - self.drop_prob
        shape = (x.shape[0],) + (1,) * (x.ndim - 1)
        random_tensor = keep_prob + torch.rand(shape, dtype=x.dtype, device=x.device)
        random_tensor.floor_()
        return x.div(keep_prob) * random_tensor

class ConvNeXtBlock(nn.Module):
    def __init__(self, dim, drop_path=0., layer_scale_init_value=1e-6):
        super().__init__()
        self.dwconv = nn.Conv2d(dim, dim, kernel_size=7, padding=3, groups=dim)
        self.norm = LayerNorm(dim, eps=1e-6, data_format="channels_last")
        self.pwconv1 = nn.Linear(dim, 4 * dim)
        self.act = nn.GELU()
        self.pwconv2 = nn.Linear(4 * dim, dim)
        self.gamma = nn.Parameter(layer_scale_init_value * torch.ones(dim), requires_grad=True) \
            if layer_scale_init_value > 0 else None
        self.drop_path = DropPath(drop_path) if drop_path > 0. else nn.Identity()

    def forward(self, x):
        residual = x
        x = self.dwconv(x)
        x = x.permute(0, 2, 3, 1)
        x = self.norm(x)
        x = self.pwconv1(x)
        x = self.act(x)
        x = self.pwconv2(x)
        if self.gamma is not None:
            x = self.gamma * x
        x = x.permute(0, 3, 1, 2)
        x = residual + self.drop_path(x)
        return x


# =========================================================
# 2) ConvNeXt Backbone (Multi-scale Output)
# =========================================================
class MSLNAdapter(nn.Module):
    def __init__(self, dim, bottleneck_ratio=4, kernel_size=5, use_sigmoid_input=False):
        super().__init__()
        b = max(8, dim // bottleneck_ratio)
        self.reduce = nn.Conv2d(dim, b, kernel_size=1, bias=False)
        self.msln   = MaxSigLayerNormalized(b, b, kernel_size=kernel_size, stride=1, use_sigmoid_input=use_sigmoid_input)
        self.expand = nn.Conv2d(b, dim, kernel_size=1, bias=False)
        # FIX: Initialize to 0.1 instead of 0.0 to allow gradients to flow immediately
        self.alpha  = nn.Parameter(torch.tensor(0.1))

    def forward(self, x):
        return x + self.alpha * self.expand(self.msln(self.reduce(x)))
        
class ConvNeXtBackbone(nn.Module):
    def __init__(self, in_chans=1, depths=[3, 3, 9, 3], dims=[96, 192, 384, 768],
                 drop_path_rate=0., layer_scale_init_value=1e-6):
        super().__init__()

        self.downsample_layers = nn.ModuleList()
        stem = nn.Sequential(
            nn.Conv2d(in_chans, dims[0], kernel_size=4, stride=4),
            LayerNorm(dims[0], eps=1e-6, data_format="channels_first"),
        )
        self.downsample_layers.append(stem)

        for i in range(1, 4):
            down = nn.Sequential(
                LayerNorm(dims[i - 1], eps=1e-6, data_format="channels_first"),
                nn.Conv2d(dims[i - 1], dims[i], kernel_size=2, stride=2),
            )
            self.downsample_layers.append(down)

        self.stages = nn.ModuleList()
        dp_rates = [x.item() for x in torch.linspace(0, drop_path_rate, sum(depths))]
        cur = 0
        for i in range(4):
            blocks = nn.Sequential(*[
                ConvNeXtBlock(
                    dim=dims[i],
                    drop_path=dp_rates[cur + j],
                    layer_scale_init_value=layer_scale_init_value
                )
                for j in range(depths[i])
            ])
            self.stages.append(blocks)
            cur += depths[i]

        self.out_norms = nn.ModuleList([LayerNorm(dims[i], eps=1e-6, data_format="channels_first") for i in range(4)])
        
        self.adapters = nn.ModuleList([
            nn.Identity(),                      
            nn.Identity(),                      
            MSLNAdapter(dims[2], kernel_size=5),
            nn.Identity()                       
        ])    

    def forward(self, x):
        outs = []
        for i in range(4):
            x = self.downsample_layers[i](x)
            x = self.stages[i](x)
            x = self.adapters[i](x) 
            outs.append(self.out_norms[i](x))
        return outs


# =========================================================
# 3) UPerNet Decoder (1-Channel Output)
# =========================================================

class UPerNetDecoder(nn.Module):
    def __init__(self, in_channels=[96, 192, 384, 768], channels=256):
        super().__init__()
        self.lateral_convs = nn.ModuleList([nn.Conv2d(in_ch, channels, 1) for in_ch in in_channels])
        self.fpn_convs = nn.ModuleList([nn.Conv2d(channels, channels, 3, padding=1) for _ in in_channels])

        self.psp_pooling = nn.AdaptiveAvgPool2d(1)
        self.psp_conv = nn.Conv2d(in_channels[-1], channels, 1)

        self.fusion = nn.Conv2d(channels, channels, 3, padding=1)
        self.cls = nn.Conv2d(channels, 1, 1)

    def forward(self, features):
        laterals = [conv(f) for conv, f in zip(self.lateral_convs, features)]

        for i in range(len(laterals) - 1, 0, -1):
            laterals[i - 1] = laterals[i - 1] + F.interpolate(
                laterals[i],
                size=laterals[i - 1].shape[2:],
                mode="bilinear",
                align_corners=False
            )

        fpn_outs = [conv(lat) for conv, lat in zip(self.fpn_convs, laterals)]
        target_size = fpn_outs[0].shape[2:]

        fpn_outs = [
            o if o.shape[2:] == target_size else F.interpolate(o, size=target_size, mode="bilinear", align_corners=False)
            for o in fpn_outs
        ]

        psp = self.psp_conv(self.psp_pooling(features[-1]))
        psp = F.interpolate(psp, size=target_size, mode="bilinear", align_corners=False)

        fused = self.fusion(sum(fpn_outs) / len(fpn_outs) + psp)
        return self.cls(fused)


# =========================================================
# 4) Full Segmentation Model
# =========================================================

class ConvNeXtBinarySegmentation(nn.Module):
    def __init__(self, backbone="tiny"):
        super().__init__()
        if backbone == "tiny":
            self.backbone = ConvNeXtBackbone(depths=[3, 3, 9, 3], dims=[96, 192, 384, 768])
            in_channels = [96, 192, 384, 768]
        elif backbone == "small":
            self.backbone = ConvNeXtBackbone(depths=[3, 3, 27, 3], dims=[96, 192, 384, 768])
            in_channels = [96, 192, 384, 768]
        elif backbone == "base":
            self.backbone = ConvNeXtBackbone(depths=[3, 3, 27, 3], dims=[128, 256, 512, 1024])
            in_channels = [128, 256, 512, 1024]
        else:
            raise ValueError(f"Unknown backbone: {backbone}")

        self.decoder = UPerNetDecoder(in_channels=in_channels)

    def forward(self, x):
        features = self.backbone(x)
        return self.decoder(features)

# %%
class BinarySegmentationDataset(Dataset):
    def __init__(self, image_dir, mask_dir, img_size=512, augment=False):
        self.image_dir = Path(image_dir)
        self.mask_dir = Path(mask_dir)
        self.img_size = img_size
        self.augment = augment
        
        self.images = sorted(list(self.image_dir.glob('*.png')))
        

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img_path = self.images[idx]
        
        mask_path = Path(str(img_path).replace('/images/', '/masks/'))        
                
        image = cv2.imread(str(img_path), cv2.IMREAD_ANYDEPTH)
        image = cv2.resize(image, (self.img_size, self.img_size))
        image = image.astype(np.float32)
        image = (image - image.min()) / (image.max() - image.min() + 1e-8)
        
        image = np.expand_dims(image, axis=0)
        image = torch.from_numpy(image)
        
        mask = cv2.imread(str(mask_path), cv2.IMREAD_ANYDEPTH)
        mask = cv2.resize(mask, (self.img_size, self.img_size), interpolation=cv2.INTER_NEAREST)
        mask = (mask > 0).astype(np.float32)
        mask = torch.from_numpy(mask)
        
        return image, mask
```

```python
# %%
# =========================================================
# 6) Training Function (NO test_loader here)
# =========================================================

def train_model(model, train_loader, val_loader, epochs=50, device="cuda", save_dir="checkpoints"):
    save_dir = Path(save_dir)
    save_dir.mkdir(parents=True, exist_ok=True)

    model = model.to(device)
    
    # Target alphas (will be linearly warmed up to these values)
    criterion = MaxSigAwareLoss(
        alpha_bce=1.0, alpha_dice=1.0,
        alpha_sas=0.05, alpha_mbe=0.0, alpha_mmd=0.05, alpha_pdr=1e-3,
    ).to(device)

    optimizer = optim.AdamW(
        [
            {"params": model.backbone.parameters(), "lr": 1e-4},
            {"params": model.decoder.parameters(), "lr": 1e-4},
        ],
        weight_decay=0.05,
    )

    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)

    best_dice = -1.0
    warmup_epochs = 10  # Number of epochs to linearly ramp auxiliary losses

    for epoch in range(epochs):
        start_time = perf_counter()
        model.train()
        train_loss = 0.0

        # --- Warmup Logic for Auxiliary Losses ---
        scale = min(1.0, epoch / warmup_epochs)
        criterion.a_sas = 0.05 * scale
        criterion.a_mbe = 0.0 * scale
        criterion.a_mmd = 0.05 * scale
        criterion.a_pdr = 1e-3 * scale

        for images, masks in train_loader:
            images = images.to(device)
            masks = masks.to(device).float().unsqueeze(1)  # [B,1,H,W]

            optimizer.zero_grad()
            logits = model(images)  # [B,1,h,w]
            logits = F.interpolate(logits, size=masks.shape[-2:], mode="bilinear", align_corners=False)

            # FIX: Pass model=model so PDR and MaxSig layer are updated!
            loss = criterion(logits, masks, model=model)
            loss.backward()
            optimizer.step()

            train_loss += loss.item()

        train_loss /= max(1, len(train_loader))
        scheduler.step()

        # Validation
        val_stats = evaluate_model(model, val_loader, device=device)
        val_loss = val_stats["loss"]
        val_dice = val_stats["dice"]
        val_iou = val_stats["iou"]

        print(
            f"Epoch {epoch+1:>3}/{epochs} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f} | Val Dice: {val_dice:.4f} | Val IoU: {val_iou:.4f} | "
            f"SAS Scale: {scale:.2f}"
        )

        if val_dice > best_dice:
            best_dice = val_dice
            torch.save(
                {
                    "epoch": epoch,
                    "model_state_dict": model.state_dict(),
                    "optimizer_state_dict": optimizer.state_dict(),
                    "val_loss": val_loss,
                    "val_dice": val_dice,
                    "val_iou": val_iou,
                },
                save_dir / "best_model_convnext_msaloss.pth",
            )
            print(f"✓ Saved best model (Val Dice: {val_dice:.4f} | Val IoU: {val_iou:.4f})")
        end_time = perf_counter()
        print(f"Time taken: {end_time - start_time:.2f} seconds.\n")

    return model

# %%
# =========================================================
# 7) Inference Function
# =========================================================

@torch.no_grad()
def predict(model, image_path, device="cuda", threshold=0.5, img_size=512):
    model.eval()
    model = model.to(device)

    image_path = str(image_path)
    img = cv2.imread(image_path, cv2.IMREAD_ANYDEPTH)
    
    # FIX: Removed erroneous .float() on numpy scalar
    img = (img - img.min()) / (img.max() - img.min() + 1e-8)

    orig_h, orig_w = img.shape[:2]    
    img_resized = cv2.resize(img, (img_size, img_size), interpolation=cv2.INTER_LINEAR)

    # FIX: Add two unsqueeze to create [B, C, H, W]
    x = torch.from_numpy(img_resized).unsqueeze(0).unsqueeze(0)
    x = x.to(device)

    logits = model(x)  # [1,1,h,w]
    probs = torch.sigmoid(logits)
    pred = (probs > threshold).float().squeeze(0).squeeze(0).cpu().numpy()

    pred_resized = cv2.resize(pred, (orig_w, orig_h), interpolation=cv2.INTER_NEAREST)
    prob_resized = cv2.resize(probs.squeeze(0).squeeze(0).cpu().numpy(), (orig_w, orig_h), interpolation=cv2.INTER_LINEAR)

    return pred_resized, prob_resized

# %%
# =========================================================
# 8) Visualization Function
# =========================================================

def visualize_prediction(image_path, pred_mask, output_path="output.png"):
    image_path = str(image_path)
    output_path = str(output_path)

    img = cv2.imread(image_path, cv2.IMREAD_COLOR)
    if img is None:
        raise RuntimeError(f"cv2.imread failed for image: {image_path}")

    mask_color = np.zeros_like(img)
    mask_color[pred_mask == 1] = [0, 255, 0]  # green overlay in BGR

    overlay = cv2.addWeighted(img, 0.7, mask_color, 0.3, 0)
    cv2.imwrite(output_path, overlay)
    print(f"✓ Saved visualization to {output_path}")

# %%
# =========================================================
# 9) Main
# =========================================================
if __name__ == "__main__":
    torch.backends.cudnn.benchmark = True

    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
    IMAGE_SIZE = 512
    BATCH_SIZE = 8
    EPOCHS = 100
    BACKBONE = "tiny"

    ROOT_DIR = Path('/home/jupyter-r.yazdi/myproject/dti3d/0pycodes/UnetMaxsig/BUS-UCLM/dataset')

    print(f"Using device: {DEVICE}")

    model = ConvNeXtBinarySegmentation(backbone=BACKBONE)
    total_params = sum(p.numel() for p in model.parameters())
    print(f"Total parameters: {total_params:,}")

    train_dataset = BinarySegmentationDataset(
        image_dir=ROOT_DIR / "train" / "images",
        mask_dir=ROOT_DIR / "train" / "masks",
        img_size=IMAGE_SIZE,
        augment=False,
    )

    val_dataset = BinarySegmentationDataset(
        image_dir=ROOT_DIR / "val" / "images",
        mask_dir=ROOT_DIR / "val" / "masks",
        img_size=IMAGE_SIZE,
        augment=False,
    )

    test_dataset = BinarySegmentationDataset(
        image_dir=ROOT_DIR / "test" / "images",
        mask_dir=ROOT_DIR / "test" / "masks",
        img_size=IMAGE_SIZE,
        augment=False,
    )

    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0, pin_memory=(DEVICE == "cuda"))
    val_loader   = DataLoader(val_dataset,   batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=(DEVICE == "cuda"))
    test_loader  = DataLoader(test_dataset,  batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=(DEVICE == "cuda"))

    print(f"\nTrain dataset size: {len(train_dataset)}")
    print(f"Train loader batches: {len(train_loader)}\n")
    data_batch, mask_batch = next(iter(train_loader))
    print("data_batch.size=", data_batch.size()) 
    
    print(f"\nValidation dataset size: {len(val_dataset)}")
    print(f"Validation loader batches: {len(val_loader)}\n")
    
    print(f"Test dataset size: {len(test_dataset)}") 
    print(f"Test loader batches: {len(test_loader)}\n")

    save_dir = '/home/jupyter-r.yazdi/myproject/dti3d/0pycodes/UnetMaxsig/BUS-UCLM/'

    print("Training started...\n")
    model = train_model(
        model,
        train_loader,
        val_loader,
        epochs=EPOCHS,
        device=DEVICE,
        save_dir=save_dir
    )

    ckpt_path = save_dir + "best_model_convnext_msaloss.pth"
    checkpoint = torch.load(ckpt_path, map_location=DEVICE)
    model.load_state_dict(checkpoint["model_state_dict"])
    print(f"\n✓ Loaded best model from epoch {checkpoint['epoch'] + 1} | Val Dice: {checkpoint['val_dice']:.4f}")

    test_stats = evaluate_model(model.to(DEVICE), test_loader, device=DEVICE)
    print(f"TEST | Loss: {test_stats['loss']:.4f} | Dice: {test_stats['dice']:.4f} | IoU: {test_stats['iou']:.4f}")
    print("Finished.")
```

```python
# ============================================================
# Compare Baseline vs +MaxSig vs +MaxSig+MSAL on ONE specific file
# ============================================================
ROOT = '/home/jupyter-r.yazdi/myproject/dti3d/0pycodes/UnetMaxsig/BUS-UCLM/'
# ============================================================
# Compare Baseline vs +MaxSig vs +MaxSig+MSAL on ONE specific file
# ============================================================
import cv2
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
device = "cuda" if torch.cuda.is_available() else "cpu"
# ---- Hardcode the file here ----
MASK_NAME = ROOT + "dataset/test/masks/ALWI_018.png"   # <-- edit this
IMG_SIZE  = 512   # matches IMAGE_SIZE / BinarySegmentationDataset in all three ConvNeXt notebooks
BACKBONE  = "tiny"
print(MASK_NAME)
# ---- Load + preprocess exactly like BinarySegmentationDataset.__getitem__
#      (note: this ConvNeXt dataset does NOT use albumentations - plain
#      cv2 + numpy, mask uses INTER_NEAREST, image uses default cv2 resize) ----
image = cv2.imread(MASK_NAME.replace('/masks/', '/images/'), cv2.IMREAD_ANYDEPTH)
image = cv2.resize(image, (IMG_SIZE, IMG_SIZE))
image = image.astype(np.float32)
image = (image - image.min()) / (image.max() - image.min())
image = np.expand_dims(image, axis=0)          # [1, H, W]
x = torch.from_numpy(image).unsqueeze(0).to(device).float()   # (1,1,H,W)

mask = cv2.imread(MASK_NAME, cv2.IMREAD_ANYDEPTH)
mask = cv2.resize(mask, (IMG_SIZE, IMG_SIZE), interpolation=cv2.INTER_NEAREST)
mask = (mask > 0).astype(np.float32)
y = torch.from_numpy(mask).unsqueeze(0).unsqueeze(0).to(device)  # (1,1,H,W)

# ---- Checkpoint paths (as actually saved in each notebook) ----
ckpt_baseline    = ROOT + "best_model_convnext.pth"
ckpt_maxsig      = ROOT + "best_model_convnext_maxsig.pth"
ckpt_maxsig_msal = ROOT + "best_model_convnext_msaloss.pth"

# ============================================================
# NOTE: unlike the other backbones, ConvNeXt baseline vs +MaxSig are NOT
# the same class toggled by a flag - Deeplab/SegFormer/Swin-UNet use an
# on/off arg, but here ConvNeXtBackbone in ConvNeXt-Seg.ipynb has no
# `self.adapters` at all, while ConvNeXt-MaxSig.ipynb's ConvNeXtBackbone
# hard-codes an adapters ModuleList (MSLNAdapter on stage 2 only, identity
# elsewhere) with no constructor arg to disable it. Both MaxSig notebooks
# (MaxSig and MaxSig+MSAL) share the exact same architecture - only the
# training loss differs - so one MaxSig backbone class covers both
# checkpoints.
#
# This script assumes ConvNeXtBlock, LayerNorm, DropPath, MSLNAdapter,
# MaxSigLayerNormalized, ConvNeXtBackbone (the MaxSig version, with
# adapters), UPerNetDecoder, and ConvNeXtBinarySegmentation are already
# defined in your notebook session from ConvNeXt-MaxSig.ipynb. It adds a
# separate *baseline* backbone/model class locally (no adapters) so both
# variants can be built side by side without a name collision.
# ============================================================
class ConvNeXtBackboneBaseline(nn.Module):
    def __init__(self, in_chans=1, depths=[3, 3, 9, 3], dims=[96, 192, 384, 768],
                 drop_path_rate=0., layer_scale_init_value=1e-6):
        super().__init__()
        self.downsample_layers = nn.ModuleList()
        stem = nn.Sequential(
            nn.Conv2d(in_chans, dims[0], kernel_size=4, stride=4),
            LayerNorm(dims[0], eps=1e-6, data_format="channels_first"),
        )
        self.downsample_layers.append(stem)
        for i in range(3):
            downsample_layer = nn.Sequential(
                LayerNorm(dims[i], eps=1e-6, data_format="channels_first"),
                nn.Conv2d(dims[i], dims[i + 1], kernel_size=2, stride=2),
            )
            self.downsample_layers.append(downsample_layer)

        self.stages = nn.ModuleList()
        dp_rates = [x.item() for x in torch.linspace(0, drop_path_rate, sum(depths))]
        cur = 0
        for i in range(4):
            blocks = nn.Sequential(*[
                ConvNeXtBlock(dim=dims[i], drop_path=dp_rates[cur + j],
                               layer_scale_init_value=layer_scale_init_value)
                for j in range(depths[i])
            ])
            self.stages.append(blocks)
            cur += depths[i]

        self.out_norms = nn.ModuleList([LayerNorm(dims[i], eps=1e-6, data_format="channels_first") for i in range(4)])

    def forward(self, x):
        outs = []
        for i in range(4):
            x = self.downsample_layers[i](x)
            x = self.stages[i](x)
            outs.append(self.out_norms[i](x))
        return outs


class ConvNeXtBinarySegmentationBaseline(nn.Module):
    def __init__(self, backbone="tiny"):
        super().__init__()
        if backbone == "tiny":
            self.backbone = ConvNeXtBackboneBaseline(depths=[3, 3, 9, 3], dims=[96, 192, 384, 768])
            in_channels = [96, 192, 384, 768]
        elif backbone == "small":
            self.backbone = ConvNeXtBackboneBaseline(depths=[3, 3, 27, 3], dims=[96, 192, 384, 768])
            in_channels = [96, 192, 384, 768]
        elif backbone == "base":
            self.backbone = ConvNeXtBackboneBaseline(depths=[3, 3, 27, 3], dims=[128, 256, 512, 1024])
            in_channels = [128, 256, 512, 1024]
        else:
            raise ValueError(f"Unknown backbone: {backbone}")
        self.decoder = UPerNetDecoder(in_channels=in_channels)

    def forward(self, x):
        features = self.backbone(x)
        return self.decoder(features)


model_baseline    = ConvNeXtBinarySegmentationBaseline(backbone=BACKBONE)
model_maxsig      = ConvNeXtBinarySegmentation(backbone=BACKBONE)
model_maxsig_msal = ConvNeXtBinarySegmentation(backbone=BACKBONE)

# NOTE: these .pth files are full training checkpoints (epoch, optimizer
# state, val metrics, ...), not raw state_dicts - pull out "model_state_dict"
# (confirmed from each notebook's own torch.save/load_state_dict calls).
model_baseline.load_state_dict(torch.load(ckpt_baseline, map_location=device)["model_state_dict"])
model_maxsig.load_state_dict(torch.load(ckpt_maxsig, map_location=device)["model_state_dict"])
model_maxsig_msal.load_state_dict(torch.load(ckpt_maxsig_msal, map_location=device)["model_state_dict"])
for m in (model_baseline, model_maxsig, model_maxsig_msal):
    m.to(device).eval()

# ---- Inference ----
with torch.no_grad():
    pred_base = (torch.sigmoid(model_baseline(x))    > 0.5).float().cpu().numpy()[0, 0]
    pred_ms   = (torch.sigmoid(model_maxsig(x))      > 0.5).float().cpu().numpy()[0, 0]
    pred_msal = (torch.sigmoid(model_maxsig_msal(x)) > 0.5).float().cpu().numpy()[0, 0]
img_np  = x.cpu().numpy()[0, 0]
mask_np = y.cpu().numpy()[0, 0]

# ---- Plot single row ----
panels = [img_np, mask_np, pred_base, pred_ms, pred_msal]
titles = ["Input", "Ground Truth", "Baseline", "+MaxSig", "+MaxSig+MSAL"]
fig, axes = plt.subplots(1, 5, figsize=(15, 3.2))
for ax, panel, title in zip(axes, panels, titles):
    ax.imshow(panel, cmap="gray")
    ax.set_title(title, fontsize=12)
    ax.axis("off")
fig.suptitle(MASK_NAME.split("/")[-1], fontsize=10, y=1.05)
plt.tight_layout()
plt.savefig(ROOT + f"convnext_comparison_{MASK_NAME.split('/')[-1]}", dpi=300, bbox_inches="tight")
plt.show()
```
